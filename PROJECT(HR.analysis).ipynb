```python
data["Joining_year"]=data["Joining_Date"].dt.year
data["Joining_month"]=data["Joining_Date"].dt.month_name()
data["Joining_day"]=data["Joining_Date"].dt.day
```

```python
today = pd.Timestamp.today()
data["End_Date"]=data["Exit_Date"].fillna(today)
data["Tenure_year"]=((data["End_Date"]-data["Joining_Date"]).dt.days/365.25)
```
